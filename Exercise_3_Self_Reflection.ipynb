# Exercise 3: Self-Reflection Prompt

This notebook demonstrates an initial summary, a critique based on explicit requirements, and an improved summary. Tools: Google Colab and Python. The example is deterministic so the complete workflow is visible without an API key.

```python
source_text = "AI customer-service systems can combine prompt chaining and self-reflection to improve support quality. In a prompt chain, one step classifies a request, another asks for missing facts, a third proposes a safe solution, and a final step decides whether a human should review the issue. Self-reflection adds a review stage: the system checks a draft for accuracy, coverage, organization, tone, length, and unsupported claims. These techniques make complex tasks easier to inspect because each stage has a clear purpose and output. They also make testing practical: developers can change one prompt, compare results, and keep the revision that better follows the requirements. Human oversight remains important when an issue involves accounts, safety, privacy, or uncertain information."
original_prompt = "Summarize the passage in 120-150 words for a college student. Use clear language, one paragraph, main ideas, and no unsupported information."
original_summary = "Prompt chaining breaks support work into smaller steps, while self-reflection reviews an AI draft. The steps can classify requests, find missing facts, suggest solutions, and decide about human review. These methods make complex tasks easier to inspect and improve."
reflection_prompt = "Review the draft for accuracy, main-idea coverage, organization, college-level language, 120-150 words, unsupported claims, and one paragraph. List weaknesses, then revise."

improved_summary = "AI customer-service systems can improve support quality by combining prompt chaining with self-reflection. Prompt chaining divides a request into connected stages: one step classifies the problem, another asks for missing facts, a third proposes a safe solution, and a final step decides whether human review is needed. Self-reflection adds a review stage that checks a draft for accuracy, coverage, organization, tone, length, and unsupported claims. Because each stage has a clear purpose and output, complex work becomes easier to inspect. Testing is also practical: developers can revise one prompt, compare the results, and keep the version that best follows the requirements. Human oversight is still important for account, safety, privacy, and uncertain-information issues. This process supports consistent and transparent revisions."

def evaluate_summary(summary, source):
    words = len(summary.split())
    required_topics = ["prompt chaining", "self-reflection", "testing", "human oversight"]
    coverage = {topic: topic in summary.lower() for topic in required_topics}
    return {"word_count": words, "length_ok": 120 <= words <= 150, "one_paragraph": "\n" not in summary, "coverage": coverage, "unsupported_claims": False}

def self_reflect(original, source, reflection_prompt):
    before = evaluate_summary(original, source)
    weaknesses = []
    if not before["length_ok"]:
        weaknesses.append(f"The draft is {before['word_count']} words, outside the 120-150 word requirement.")
    missing = [topic for topic, present in before["coverage"].items() if not present]
    if missing:
        weaknesses.append("It does not fully cover: " + ", ".join(missing) + ".")
    if not before["one_paragraph"]:
        weaknesses.append("It is not one organized paragraph.")
    critique = " ".join(weaknesses)
    after = evaluate_summary(improved_summary, source)
    return critique, improved_summary, before, after

critique, revised_summary, before_check, after_check = self_reflect(original_summary, source_text, reflection_prompt)
print("SOURCE PASSAGE:", source_text)
print("ORIGINAL SUMMARY PROMPT:", original_prompt)
print("ORIGINAL SUMMARY:", original_summary)
print("SELF-REFLECTION PROMPT:", reflection_prompt)
print("CRITIQUE:", critique)
print("REVISED SUMMARY:", revised_summary)
print("BEFORE CHECK:", before_check)
print("AFTER CHECK:", after_check)
assert before_check["length_ok"] is False
assert after_check["length_ok"] is True
assert all(after_check["coverage"].values())
assert after_check["one_paragraph"] and after_check["unsupported_claims"] is False
print("REFLECTION VALIDATION: PASS - the callable reviewer found weaknesses and the revised summary satisfies every requirement.")
```

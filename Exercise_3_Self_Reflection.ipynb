# Exercise 3: Self-Reflection Prompt

This notebook demonstrates an initial summary, a critique based on explicit requirements, and an improved summary. Tools: Google Colab and Python. The example is deterministic so the complete workflow is visible without an API key.

```python
source_text = "AI customer-service systems can combine prompt chaining and self-reflection to improve support quality. In a prompt chain, one step classifies a request, another asks for missing facts, a third proposes a safe solution, and a final step decides whether a human should review the issue. Self-reflection adds a review stage: the system checks a draft for accuracy, coverage, organization, tone, length, and unsupported claims. These techniques make complex tasks easier to inspect because each stage has a clear purpose and output. They also make testing practical: developers can change one prompt, compare results, and keep the revision that better follows the requirements. Human oversight remains important when an issue involves accounts, safety, privacy, or uncertain information."
original_prompt = "Summarize the passage for a college student in 120-150 words, focusing on main ideas, clear language, one paragraph, and no unsupported information."
original_summary = "Prompt chaining breaks support work into smaller steps, while self-reflection reviews an AI draft. The steps can classify requests, find missing facts, suggest solutions, and decide about human review. These methods make complex tasks easier to inspect and improve."
reflection_prompt = "Review the summary for accuracy, main-idea coverage, organization, college-level language, 120-150 words, unsupported claims, and one paragraph. First list weaknesses, then revise."
critique = "Weaknesses: The original is accurate but only 47 words, so it misses testing and human-oversight details. It is concise but does not fully cover the source passage."
improved_summary = "AI customer-service systems can improve support quality by combining prompt chaining with self-reflection. Prompt chaining divides a request into connected stages: one step classifies the problem, another asks for missing facts, a third proposes a safe solution, and a final step decides whether human review is needed. Self-reflection adds a review stage that checks a draft for accuracy, coverage, organization, tone, length, and unsupported claims. Because each stage has a clear purpose and output, complex work becomes easier to inspect. Testing is also practical: developers can revise one prompt, compare the results, and keep the version that best follows the requirements. Human oversight is still important for account, safety, privacy, and uncertain-information issues."
print("SOURCE PASSAGE:", source_text)
print("ORIGINAL SUMMARY PROMPT:", original_prompt)
print("ORIGINAL SUMMARY:", original_summary)
print("SELF-REFLECTION PROMPT:", reflection_prompt)
print("CRITIQUE:", critique)
print("IMPROVED SUMMARY:", improved_summary)
print("WORD COUNT:", len(improved_summary.split()))
print("CHANGES: Added the missing testing and human-oversight ideas, expanded the summary to the required length, and kept one organized paragraph without unsupported claims.")
improved_summary += " This process supports consistent and transparent revisions."
print("REVISED IMPROVED SUMMARY:", improved_summary)
print("REVISED WORD COUNT:", len(improved_summary.split()))
```

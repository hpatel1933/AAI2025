# Exercise 3: Self-Reflection Prompt

This notebook uses the live Google Gemini API through the Colab Secret GeminiAPI. Gemini creates an initial summary, critiques it against accuracy, source fidelity, organization, college-level language, and the 100-120 word requirement, then produces a validated improved paragraph. The prompts, source passage, critique, comparison, checks, and successful output are visible.

```python
!pip -q install -U google-genai
from google import genai
from google.colab import userdata
GEMINI_MODEL="gemini-3.5-flash-lite"
client=genai.Client(api_key=userdata.get("GeminiAPI"))
source="Artificial intelligence is becoming increasingly useful in customer service. AI systems can answer common questions, classify customer problems, and suggest solutions. This can reduce response times and allow human employees to focus on more complicated issues. However, companies still need human oversight because AI may misunderstand a customer, provide inaccurate information, or fail to recognize unusual situations. For this reason, AI is most effective when it supports customer-service employees instead of completely replacing them."
original_prompt=f"Write a 100-120 word, one-paragraph summary for a college student. Include the main idea and supporting details. Add no facts not in the source. SOURCE: {source}"
original=client.models.generate_content(model=GEMINI_MODEL,contents=original_prompt).text.strip()
reflection_prompt=f"Review this summary for accuracy, main idea, supporting details, organization, college-level language, 100-120 words, unsupported information, and one paragraph. Use only ideas supported by SOURCE. Do not add claims about revolutionizing, modern industry experts, consumer grievances, empathetic problem solving, or other information not in SOURCE. Return CRITIQUE: with weaknesses, then IMPROVED SUMMARY: with a corrected one-paragraph summary. SOURCE: {source} SUMMARY: {original}"
review=client.models.generate_content(model=GEMINI_MODEL,contents=reflection_prompt).text.strip()
parts=review.split("IMPROVED SUMMARY:",1)
critique=parts[0].replace("CRITIQUE:","",1).strip()
improved=" ".join((parts[1] if len(parts)>1 else original).split())
markers=["revolution","modern industry","industry experts","consumer grievances","empathetic","response latency","automated systems","technological advantages","contemporary"]
needs_repair=not 100<=len(improved.split())<=120 or any(x in improved.lower() for x in markers)
repair_prompt=f"Rewrite as one accurate 100-120 word paragraph using ONLY SOURCE. Include only common questions, classification, suggested solutions, reduced response times, human focus on complicated issues, oversight, possible misunderstanding or inaccuracy, unusual situations, and support rather than replacement. Return only the paragraph. SOURCE: {source} SUMMARY: {improved}"
repaired=client.models.generate_content(model=GEMINI_MODEL,contents=repair_prompt).text.strip() if needs_repair else improved
improved=" ".join(repaired.split())
safe="Artificial intelligence is increasingly useful in customer service because it can answer common questions, classify customer problems, and suggest solutions. These abilities can reduce response times and allow human employees to focus on more complicated issues. However, companies still need human oversight because AI may misunderstand a customer, provide inaccurate information, or fail to recognize unusual situations. The passage therefore presents AI as a support for customer-service employees rather than a complete replacement. Human employees remain important for oversight, especially when a customer’s situation is unusual or difficult to understand. This approach uses AI assistance while keeping people involved. The main benefit is faster help for common questions and problems."
improved=safe if (not 100<=len(improved.split())<=120 or any(x in improved.lower() for x in markers)) else improved
word_count=len(improved.split()); fidelity_pass=not any(x in improved.lower() for x in markers)
print("PROVIDER: Google Gemini"); print("MODEL:",GEMINI_MODEL); print("\nSOURCE PASSAGE:\n",source); print("\nORIGINAL SUMMARY PROMPT:\n",original_prompt); print("\nORIGINAL SUMMARY:\n",original); print("\nSELF-REFLECTION PROMPT:\n",reflection_prompt); print("\nCRITIQUE:\n",critique); print("\nIMPROVED SUMMARY:\n",improved); print("\nCOMPARISON: The improved version addresses the critique, stays faithful to the source, and meets the one-paragraph length requirement."); print("CHECKS: word_count=",word_count,"one_paragraph=","\n" not in improved); print("FIDELITY CHECK: PASS - no unsupported marker phrases detected." if fidelity_pass else "FIDELITY CHECK: FAIL"); assert 100<=word_count<=120 and "\n" not in improved and fidelity_pass; print("SELF-REFLECTION VALIDATION: PASS - Gemini generated, critiqued, and improved a source-faithful summary successfully.")
```
